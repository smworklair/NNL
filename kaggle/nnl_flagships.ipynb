```python
import json, zlib, base64, time, re
import kaggle_benchmarks as kbench
NL = chr(10)
D = json.loads(zlib.decompress(base64.b64decode("eNrdfdtyHOeR5quUeUXONmCcQTJiA1dz4auJmNiL2Vg5EE2iR8SQRHMBEFpwYiLABk9a0gRJSbbGEiVRskez4fWo0OgGCn2MmCfofgU/yWZ+mf+pTuimJV2sZZJAdVfVf8zDl1/m/8+XbtWqd3ZvrVe3dzdv3qntXLoe/Y9/vrS5Qf9emr9UiS6N+qN4dDZq0t/9UTJq8wej34+S8bNRe/x01B8/H7Uj+qA/fjRqjzrjJ/RzUolGx3RDZ9ygLx3QlXjUHb8wXz4bDei/Pv05Gz+nr7bGB+PD0Qk/vfS+Y7qjSy8545bQfU1qBd05fljwBfpgMBrScwb01cHodNQaxeOX44/5ReOH4yN6wvj5+OGlf6lEpssLhV3+lpp4jFZ16Ce8JRoN6NcTfI1ewR15Pn4sd42P/McuTvdYOyDjx5l3JNythJpPQzTq+S9ZKnzJF/RVHofY//py4ddfu2Ecv6hglHim8C16Jw3ikT/4g1GPF0Bv1KNGcytb1J9H9DEtBWplHNEyadMDeRZohYStWClsxTu6m55CN1Jvud/hOsmuuCGGsZ2d5SEuJ9w+utzG2HFT+txavm8graMu9vjbh9mxXS1s5pf0U0eWsn/D1cIbfoeX0TDwsPhtkTXc19U6GB3zjGGzNM2aii5Tf3k992QfFT6JF07Jk0bnEY8rjcgjupjQkPGXz2kQO7pd2ngarz18gnXHn7Qy7/F7fe2n6fWP1lZ+TETXOnh0e3TuN35+rrD1n2rDfpx5a5U/bYK562OpN7W/vkAYP8dW6fGKpK2CVQ0posJxgM3YCfo9P4kk4KeaAU/yB1/flJZXdKXBMsEX5fQySI9n6DSNEMQ5P68rMsLsYF+y8L3n3Le2kbLUsGCbzhfL7z9jWycQBbqM5HE8tMfBQ4ql9aehpvIHZ/wb+odntgPF1c+dmwovzJh+pysYv3Od+pjFEKSlWfJJziDJNPbwaxxh8K08C3pQrAq+srOTyMqjwUZvhmjp49TlouYGbyvWJH/WMTnxd4p5E2Tm9Yi6x/M4GD/C3OhqsFdpzjA0QatI6aAlg/GRXT7HKvxFkLdLRtqbrKBhMTVBVAq/tKmX3XOCTpcqLpqjITcPyrKRHuq0Kptu6LnxHazaE6PDeJ39Bo0WTcbPTYLWFuuvt/SQJr8h3U7YT00a4DN8XweDRj7SNvHCM9pfbC3p7YAFNX3eywxgyRoq1pffFE1jF+3t61II+1usiL6ebrBFlxRuWbFpWr4UY2NJ36n9P+QvypDR9u3Kl1kg+Hsh9pcdy9HC3b1QrKe+IyHjWzew1PjVDyFaujJW4xfB4+ZLtu8QBlhb23mke8ddbWTFyIEIPO51aBItFAvnN9wwemyMSW34uoL9A1idCYzIRMW+GJ9WSaRk8RPuP0ZcHQixCy5UmzC4h2gMP80YEClrfmGx1GjVRzqpgnUDW1bM1AHadeoruXITNdsf7rvXH/wa9OewAjmhqzaGCOV/e1b1tdOeFw+ziJQ2lnxqTy0slfSa19UhPUGmXr282E6LZ8qdGGsbs3DCBjikx9AqTVwzBkwgJRaWS/wo7l1TZnLad1ww97PR6Ou86/5cttU0e7+lR1tU7J8znoWgyyuTGmfi/rDPJP+abXnh64PXrU42wkUvS8R4lZcZ+2/CXVWsAt5gd/Q8XU2ami1Ccf4GuMlZl8Was2By4GT+FSPo9S5SEXtgFlnuNgj6fW2yGa5EeOVjODbPI+1FU9Qv4I2BGPyR+Nlso/PQjP/3+CWGCtoFNuhD695D0xiPl32qE/oolheOn4RrY3Hux24oRrfLA6umsX4g7xZjmZ5xSLd0AOg8kaliu9huY1EXo5g+fiyIhOo69sv5qkxTRyWCcSDTt/JODno7X2JDNzCEtBD4FdwN8oJoucRwLQOpuVis9D7hV1PnOmI5BXctTrrrafWNn42fuImjzuQ4DjDPRNqfptWLrNMWxubUuRr6wKBZSyVmZOaVbCE+whQqRoP5p9+0xwMgNU/SturicomegdUBWzUWVYpPT21XWO8YNwIaTvzQni4ImAbis4p/LVvUOR5BQ1Z+coHkOctx0I6pJUpFXXxjm5tJeIz7+oE0yz4l6PbqpGtPlxkNv91Qh+yZE9rRjMT9YewvMf2AmwEY1o7bleDNVyd8M3VSfL0+TBb7XvVD6OMnGBt/r6sFLs5CCoRcvDbFusb6gcdBLidBtuMjCzwFiyBY1EtzZWIAzTeL+sT8jt71Cha7usHy1aAzS/M//bL9WRene4pTJaxXDoNeL5Q4kPwQ9cQZMUoZvmhRs9BV7yusXPCNlOfxEN9u0Xi+9zZYWixZK4lYKXisP3Dw9I594D54ZLHg/hxLu1f6qKBbwXOXJzYJcnqdI50P9EkJBlpgsGGh7Y2Htu1DAY3GLFPZB6lEePCJ+iTOdRIg+AIbNsYYSxDqPMC4l1ZKpkc27IHsl+CucgTmHKLpCZy0jkqVhkLamI9YzZiMllr6GczmlLHbsSE1hafZZc3RWyYg08fABtbG+0mNoN/XJneMBgb5kP3MDYzx9I46inLNguUMRJyxZFWDeJC165bnJn19krtU1RXGKos1XMkb8TxQWAKb8xRgAZOob6eaMf8+WHVp3PJ9xdZysQR+LXKeUTfreHmrOmX8Li/+nHZXzgD8fOot6PbExrX2LCsoIUS4ASZaq/05Q5B5ENp5y8uTWlvwFSD+eXf44TPv0RX4kRj8oaJpQxvEH4oz5glg/ykC8YmI0BtjCeBh1TTEi6ObX7oA2Sk2BjWRxTyB1CcquIPWBd0tFtvfA298idVBC2IGICQsyFwtpc3VSIEwGPC7gN6xF5DOU1w0SE1QJxLpeck25f+zwSIyNDObDlwZCBIV9Hd1cvmYQIA3xSMDbmDpHTJx6fil+I6xGAQaomGrREeQF4Y8g/6+gmaSnqARfoaw/fP3FjHFuu4HXPFQ79CAkfUCiBKgwJR9CBpx7b1EbtAAiQFN1YRK5olDzFBb7St5/lOnPng3khygxx15H8uixlSTifGxYnMGALLaqhVwNEqcF4Zu6OHBCK3M/+wC3K3h+Gd1ooNuL0wcR8cqd7ZBS6/YOe8WA0ori/+funeZYQl6XawcX8HcZwO+oXon1isDKJKGQk8BX2jgjFFhHuSFn1zsgHE6yCuPjOD2Gb/JIJ6nfjDQN5wd+S2v77A+GdEI57pYRf9RI2AHcHBKPFLjQXlhCh1rbZkyaRClVN380KKoTVHtQFaPsMMCr2ZlZRoj/AyP64ODYhTL5PGBldVyQJZnyNxpwoY5SjlQ3SJf1TEQw4OnXpR/8PJizfO1OBd2RAewdnLC/DnticwdIdK+Uha+PsDyBdtBPULDwSxwsvPeK6EBVfH+SvC4fi462c/anwHSxe++LND8Faxw3n7QdX1aNd5jBt4jwgFenZtmdvPaH+yc1WIV9MeMVEpKwpXDPMbU6sI0pvqh2rIJbuhh4ScCIx6QzOjL+u/pi5qpmAwNLK+flrZaIHXaS1dgYCbKrDQfkPRi40qxg0QJuudXKu4FHTSiyFJyFmnQ4SniEYQHOxjF41ZmuWHvaQWuLk07t81iekGundafrANBoy4mQfWg1GX7BobDajmi1FdimGnzudE5HJ8hYgYCPiCh0NVTc7uZcOFBN2EfJWl/oy9m/08xOOnZDTq8Oj3bZ6C2nUMQyFHrSXtU3Pgx3WIofvWnB8zsXrMD9GMEjf8663S1DC9r+x03JCHfg7XuuPTEuvZ00fNCNSp66rzVgBI9V6JD5Z0W3y/A4p2RMNAmxM6c87Wdp9+tdgMPHR0dhoDT1fnyhuVhz/KgrgYym4UGTiXVhAZY/gPdkHEloweAUCYYQ/yeq6hpc+U0IXyRjQZ3xRAOerwwGc1DqEVmtfVzExJOjKo3K8NDWyvO8c/7OBJ0CQ2NLpfP/lCFFfV1BqN0YqNjAw0yP/QjRKkOL045xT5r4a9x164uldo18o6us8sN/fkgxTDveWFPI2f4IhQ7HAen2DGkPW3wsc8K9Ya0Z2wQRzfTB0JodyCsnEBRfti5b4MFd1eCXi//9DLWrHZRBEyDMLskj3Zt/StVp21fVxjMMKlMO9lBp1dKQzwtba9QZJ8qXQG0wLPUUIfobkgIu7paZmjA2GyHkv/q1WlZG1DXv6ab7lQ/Wt+t39u8GeSF/eOD5cX1+zdv1XbXP9yore/Ud3br+9Xd9Zu3duv0W/3Gzq2bt6q7JUZNK0XiZnUDN4Ot5afj1zxBEA5P+XOZPfZ5PoZrQV9tCl2OW90xxGNPKLLB0lL+0LGJoMujOcFM/Rz9lhJKWAofucda//5YyK78jMAMxSg8qN7bpmHY26892Kiu36vv1XZ2b1dLkoV0KUXCOEMrOPIgvWIxyPJUNf4TZnFiWXQNvM9gYNejHuexGdCwe9vUlu3N3fXandrt3e361tb9/fumgfdLEvtIkDJEfQbvQKyAttgD2ua+GDgsZ46YY9pVZXugoEpsg7OueTQ1nwhHVDgLLIvowV+O/jD6t+yg3qZldG9788H+XrW2u7NfXd+rP9iu7uyu72zXb2+WtN1DZBLPi21cV3KlCpNGNH/1LwefLM7pdj9h2DuCk/WaiZje/eI3BXM2UHEhbu4LSTEDgVOI1gfY3Sk1gH7Vd/bqN+oPbm3UtjZr6/Vd08niHmH1Qz81NYMHSYyQQO69QX8tm7EvWpMlWBeRR/YNhRvMcD7kUFtikKJ63vxy9GlFe4FJOgVh8JkmA8jukflV0TrEKngBjh3Yv1DIWCzZWa3v0vTdvHW7CtFxo3qBv2+GMWbF00XcBEIXkDV3HpkjEsCNK4ExzbtqiKWmtw+ieUR76cFHMEiOdOa7ovdHp5AIhxJNCkkDbLHxvxVEKy2Pum+ul/Rzp3a3tr+1T1N9QyRk7c7O7t4UHR+K7OM5ezFqX5eQgdHFFrI5Tpnv2IAC9cUVYbccY+ZoBGagAJsaympVIqUatUT4ZNPmNLuxV5EgUxvshL5+Y1A2yberD9ZJ/uzcuf/g1o0qC6SbtzZrU3b+ULYaLVkgXyyxEwEGkFl6+Vf/zeBss1cqdqU8xB6wfmlT84I0N0IzCgRhiYUKiPUQw1FPQEs0V1WjYBZgpmWxQXR8u3qjvmuW9voWyazt2l0SXfvVvRKR9RXNjCbYiHTxNNZA2wFLxe5qrFIRQkNFaZMSlWgYcY910R450YHHncGnHoKJ2shTI6RD6hv7NK/3tkt78nsnR5se3SPRmNIgvOJJrLVIMU38Fhm5Qpv0hcGlmREH2IPJwa217KK7sV99UN3aIjtk6vE+ErkPZyRvBAMF7Cs1XpJWxHZFOI26HpfcXE346lC5/bQJ13J0NauBrSotl9v1u5s7O5v003Zt5xbpif3qBVCwiH3d9dLsBIlZCYtI0L4H6Bo7ka95AT0WNBJ964CPa1hwPstxqGmMCQRiW3BFNn9JoT6Vt70ipf5Owrt4TtBrLEr/oQokR5jSlHq8d29lxRce9+obu2S+qKost1YMuD6AaFIHiJ3XvuFpGrmQVtB2stIKtu0nYRqCa+RbYXYM1SwSez4rFWQSSQPoJNdgw1BP79a2qmWMRX/UUiYNMspO+YO2BWo0aHcE10z86xfSr76XOdwuGgnKcHknCpEUTUfo4NbZo7fx9IdzuZadPjLRyBjeIINTzBtIjDLj5lDkhcgyeMLsCUTO+pDh9ZYXWmzHPrhVmPgG4TnjpxybtGQROeJkNjKThBkRNbW/fn+vfmerdNGFA5Vu7DA1WywmkowRCQ7IQABXY0KKf3hgfMOsVVzf2t2u3iZno26Uiza61N04YTXnSXsT13PtpTcPgb6rFyJZIw0FVRq6a72mweroiw8ea36MfM6TlLndzh+XcXCM7fQIZDq8VZMObt5Z36ivL86Rs3X7zj5r2LLpeWuJ0bBfu2L/Lc5F4sTBXFIN6u+A68bXhPXYkrwTycrqqRwZlDuH4YZYXller25V79CmZyVa21i/W9tQ6V7bn0SixYKEyELS1Jljg1QFxQXoYstMni2CYXChQCUIxyNPz4Ugg7R/c2uPOrABvfqgSpJ5q14rZVkXmZlW0AZXr2eFq5Wkmb3OfcvpEGdBTtqdrdqD6t7mzuZdVrR7e6VLCDZCE7sicRVP7OTPIEiF2g20BNQnQrqlpWczqhNfj6x4GuiWUTWcuPWWNdqU9Sli92Ekd5xhaYj6tRnE2V7SMqvv3K3vbuuyUzlcL5MSCk4JnDo+MDveCFDJ8bYC1Kf2ttIyeCALDCYClNN1EX6+GHI2dc/kRRvJEzqlL7Pd406RH725URNvqr5XZW1zm5yrD+vbbDjtXQBueJS6pOC10X/+afTmP7tk8Qw8OOapKxlzIlVuRv0r191YiBWRWN+sC7nW0LoWpiFBj1bnrwHtUAvWQWlb4krs1i7EBjyLNh9Rw/sTDdshGmchNjEc1ETkhdnINo8HnMCt9U0BuUpg9AB6g/BV7p1IXHzqO9aGOuBBYXxvHgx2PYJz8b6gofGQ0jstTCG5Xd0RW4Cszyr5quubO2SKbrLneu/+zu0JDQIleusC82w0aRa157nRuhbLFPjEmMeh8TcMrHzFoH2RjnxLK01UdDjkTuwNq50lcjyUNPNRfy09AGSm0hqsbtX3xIxTiIr/pVX64Fb1Dnbc5jQ2hyYJGC81BhOFtpkmNXV88M0MBbx/3yV5OclQZLqzU99gNwIyorpxd3Nrc4eMqN3Nva36h3VM8EVQqFlwns1EtgFoHH0fJbIVQ86R/gpLaJImuw2QXZ3+xHbEdmL0Tp8qoVa0KdNvssWhBDCHZJJXZWVv1AmV2YfInNi+xXyYMIy27AC9FQU1bW+twGnB/jsV1ndLFoJFbXM7ZbypnfsA23fu1DZIBZT5yN/Lq47ha7QDd96uNGk0bs/6OZN1qq8WQ8dGonth6+vVe2R/mNgIOrH5gHxA9vCNiC3sxL/rGnukIF2DSYtmXfRVKhpRxxZvPhQ0iML4COS/Fbta8EW55jZYAtMzBU2FZDX07f7tO/UtA2uzCXKDJwei4v1mR7EVDfo1Cs1fRWAgO60Z0nUyI2sZOoR8Snly//b2P4YdvQi/l9ByDKq1Jw8F1lTalnXy85B852EJeGICDzk+OLlM1wy4sFfbZcW1RZEp+WVi3ZVdNeK5tsUPlHZq/kLs2RjsWR65kQ8UcYFOsk9qGLO4Ec6MCXXBwBQpM4pD3GF78x6U9Pzq+p3a7gVImX2GtdmErRELdOeykoZW5At40pEGpkwpVPtQw4O33Pyquph+C7dqDEiysDKxt3LP1fSctl7XgAlBKI2jAA60068/1bQKONWxKIWQx+UmA+vvN6mNF0paMaU31z+sw/crA30Dw1lYGQOBOQ309oqm/DX9eUN/PqE/n173mqKDb4HqCyCoPB8oGO0q+zpiuJAF7QmhslF3ciPxIpwtSaqhK2nCQLtEFOkTWuLjFrgWn5Br8ZeDT739IazjZ6nekL748E51hxWedXBKsIO3RhOgcoOmU7H7mGqEgACpibruAzeDoK+TuDUGnwD/sFvo8Gh8nPCri2Ac9dWaoCc8lTBuDgKfBBHcgYGqnipl0gY7zHNswbRwxd9mI/9WfWNz14Np9i+yDbN4ZB5CcVSEUIQ72QAKgYnE2llyU5MJlr9nn9dc/MDEFDYv6k1q+nIB/ItV5nXprvJxjKv2sRSsymA3xvZVayw9K7ShN3fIHUa/blxEWxAbwIcy89SSvNF09JhXUVHoDBGFlqFbtCAj4nT9tPvsUFTZtL5PBt1ufa/UHFCeCaw0gyeJ02ChxJR9Ih3CZryep1I9M8GUexO1GlZ5MYXfmNbzP+/TNtysbxlWD/1DXIhLu7X/tevbZfwkQeTU0hfPDVLZsUxdCTshmcVqy/Rk7PDFptIgwdXTKpDyhkcyXQxLPIQTyJbSGg9hdbtWRWt4pbVxZWvno9o2Gv1rmQD6cd5vuFTMsNVIHppaoSwGlcgYX5mNbu3u3rv+y19+9NFHs/v1+7v3b9Rmb9bv/vKj6u7NW2t7//XGr/727/7+n67e+9v/fps3wCmAL4mDkWdLAa/fUbzkT7MAT8CAiywVQ31bxxMZSDC6YkOAxkSlNcB2bfuK+98UvV7we/07GOZtiDhjZqb8bOv0SGzumQk69E2k/bIu8RONPTdDpedno7fDWJKiz09RXGfcCHth0c9UTwooIa6Di34HU8j7eQB7OR02GwF9OI90nz1Jk1m6/rhQe+cX1GSrWKUf+3VmByacGLGUMJynVOggzk59S5C5gG+CzF3jImpYGK18LCQLjl1G1vFRyFR/U1zIuF+HkUT+NfFMVNxlE9hwIwVK6JVI8gNmZ2f9mQFSrehu4Spb8ifhBxtWUZH5DZVtfDd6q0P3NQnhr+lVn9O17/H3F6PPyCb5fPQZvv316Buenq/otj9Ef3n024WrSzxKb0Z/wle/o1gyfZHv+ZaMRPrhh4jufkWLm//+Xp/1Az+If3xHv3zOt3+FX9/Sv/KY39K3/w99+KVt4BTbatnv8NfifHIgyJEFvrwwBqG8Tg39S14qzLAhArOPUYPar1IFco3v9/J6E8zHKeYe6BuHuYEQbsI0InPF72Uar7OrvEltPLKg3VDjyg2hhQebUUxnC+hCbRn553u5IpZsqoPgtV7Rq7VIibKxWBcpAwsekW9/FNkU4W617MVMg9Q0K24QCZTPzLwQEiIWkbJ3GeMRrjSRy3w+HOfCiUpTqhxCOcc5vDnSRBrqtxyRIb/FJxaqL4rsTEB9qX18wVSv+lPtufJdYzn3lPUU2PLW0U9TjNy8W6kqxV9OTHwLeKmUS+r7Qcq1qZT6Vb/Z71zk+Re/iH7BEoDXGu+H54JgGSeB9xKNqVqDPekHT5xOsEms8flySYjOi2nVBCe6Y1JOnilgGIb6fRU7Teeu+Z1jAUaikcUXNZzly59JeH0xQ5e+uy4C8xOSat+yaIS8ezX6d/7hHeTgv9F3X/En9Pu3oz/x5XckUL+gi9+O/i/9LEI0oqd8S/d+hi+tqJilTz4b/SsL21f0mUjQb0efrwnl8FxCeMEv05hlgUEpVR0HSCJuhM8PNTI2QU/sSLh8+AZm8JkIzryK+DLtxtZohSTYcWMqc3I+nJ20V/DBJZRqY55J44NLkeNJ2wjVUUA7H7+cjXIeY1OFOiZVw4IO/IVnppqsFKhE/i4781pD2jByDoWtilQdDqW+jjyygo/XUht+sE/XpFZTDUm2iFClYUHxJcuXmBV6XiCE4TU5F9TDxjylwN7II2SrHMC9jWgV9xE2QIULMS/5z2yUFmnlRuMEiQfedAaGstmuRwENTt1PxgtO3SYH8t1kSI1UC2mmN1rQ1CK5CwZDiTM3SW0sU23TFq1RXmrMnFS2gqB8FLU/z28CvxiD+MLqVZBaJf5s7XfJylSPmmxAFhl/gI30NvJ6nYeMWIXtXqvzdcZ7atagKv6IzdKcGUN1On00vxhK9oZYsGpfubTGA4dBKIATAa0yp0lIvSYand+pCOhBXoD19irwUiE0vtEyq0PZW3LQy2ss2J6aN32w2A9QxohXfapt+Q3A2NHTTnwwXEBmqXfEiK1w6uGA9PwucYfpRa3ZNc/M0HquDOYKwHgiLgayQXnxTSXMltKWnq/+WxoINtG7ExYpj+YNITeyRyfhdCK72vJvk/oC1Oz5SpQxNIZ6skcSBO1T0JKI9U6EEvkckNOCTkWU9OZ0MEFg2jt32DMUh8GhEn3F34Co2gy42AeTLs8vzUlfrs1dieSOJJehZfkjJyZkB6PeFVDVEJvi8t74eU6n46haRtQAK8iqz7W1tUhVtlZrbKtTeRIkIlO/HoILrKVBqMMtmeGQdIYtcCozn27/0ARsDdQdSypn/lC5V4rppev/KUcsDMQvJexZvR1Z1weylHmEh8LtDWb8DLQlrFHOj0hrC6qj6c3/Sg7ApRZiAKZ5YPyaaG4R0XpcA/Bz1IrrYvX71qJUkYiW5ua4/b1K4KbTQy3iEvjyjp3vsi1mfBMiAoxwqo08QjUWIQBjSnqhCTVrkKa+TkbPvpuXxoFZPsB2jjQNlMdd6nMcuG5ytoDOLIy3RmSIHLamxZTCP/BGXnG+t1orsWjhLAo7m4fMpswgmD7sCOq6H6iohzqMw7dQX2bZJB5kueqGmHqUsk8DPEaWuCW0cvjvULROM5VRRHfN5kC4gchJ8bj5hjNlFcqRapGtsvmxHiXhDZi4kpMaT940BN6V51GHLo2iaUI5oZc+Z4c7WlhxkBlZR0dC6ISNtzA3f210IgZDJgHiRAsQAWEz8SMb4Uzl2GDqeOy/QPZHbMlrgvxJeEsRv1fSaj1BA/lfbT8N0ATD/NheCjKoOHFjuBBRIN0MnZ77pdS/6WzWi9OivOkJ/MOcZRUEDtqWp8lWvqxQWyqD82VTgGCQgxHDKC0IEL6pOG7MADLOqPPQ3deXT6OMFwLn0A+L9HNyW4viMzYRNSCvVnJjJGtYzLq3DWnLrAbkwAnpRSmvHSx5y8OVDEMrKD8QirCSl3kSyBP8gMucHsIfw2kF7Bxy+K4DScpFH+LJFkxOkNQbucA9fWu4C9gPL/JKgUhakDkCBvOSoDh1WmnoEXcCfbdMYqejR0z/iooS/u0D4K1KxVvjqB7JFyFSjChVAWerAQhI7d1RKbJHFDhQ8dRBg0wIIxGj+lA5Bs9FlB8re7+jNaqNHuSbaFC+ZxPVSPhkOm23sJAPYiWaKIUjNA3UJsdlivZqej0aiqHRtAWEe37URqTVbOTntbEHaqOHXtUBR1/s6xFpfsyR9gev3pYfQeIsWt1AzzOSWpK42UT6yjXenQxj0ystH0JsHdtopTd5SKNPwYDR3dUEyDjT/5bO9bkD+ww8y0oJVrQpAsHQqav+pvkpYFG64tXWKUuxdLMjaJRI7gh+C4sLIjoAvCpTLpzAR7ZEiOcp953lVeAvuDQKG3w2Ey7njsVolq+YaTx951Pq7vZNfUATrT0PEC3srjzLtmlJ3zTUirQiWVq4G+8Lly4spZXFU2OUNGz6kbCetIGnqbT+M6tN+jb2rMWKjzRBTE28rpxZYw4FOEYVdmUnH6Y74ac1DeSIUgjAc+N3vVHTQ4+WRZKdPZbKM5Pk6JUTWfWCjz3ToHHT1AIxZevVQ7X1hbQMl/2aLO4ZDSABvBYuWdjybJqaxjI0GK9DdGDi1xYrcmeXNh1NKLOF+evQKTaS47m37jA6t8V6eibLlHDSwvJEXt0a4M9TF9rw04BL03/zOXi6PeQIOlcG7dSjs6C27qhDb/5c++0IPIByUztqDWeMpcweF9dQtC6sEta0ubRpYk5i2VG2q2E0fyg5/172Xzay9J4WSuBo59RyGBim2ROhbHlYMhZNyuWW2Zy/Zt0OLQVR7HdVciL0/NqFOfsM/jGREDgT/Ly0+Kv8wUu+hOQYbduxclylSABEZ9OblsU5exeQduMlmiNbAE5WPDDHy4hK5Ufkjs97OxtpVsVC4IG/yaWlBRVO5PRta5Q5I2KAuv1tw8NrevFCuhdiiOey4A0SKvHPZhcgpOGQRd6+p8AZlI4RtCoIjk7ZrkjLhiX2DPMj8aV7zndI3xTysxznPuRaKSBBMSJTb5U8g1yHZG3WG5spZV7gvr+mwVFij/WhEIQQUzabFn9Z1LzSP7ChCvmdkSgWLylezOTQoTQ49hXZu6o1Kro1FYRKIY0VP/1R6nsp9gicmApsTLTes9w7b5yupThSBlc5lqp59vT6phX9tkCQDl20uGzThjNFhcQMX1ie0XOo3DGcxtQX7dADZJEoqA1TlAyg6L/MpiV+4p0DGe4YW2Cw68iwFT/N0nGgJPNO44ZWL/hJjKZAGsegBn5a8Bu977gIbJlikS4GHj7Ftf+D3iZR6lejL2for2/IISc+zwxFqn5Pv7ziWNXo1awwbPqGLdXT8E63pD5HYuQt++PmUKynXh0Es9R6LqlEkgv044Y1eVLSRO0f1Y8qryIwSmlTsag3dQcUyEPWsIAL9Ol32eBCSY3D8xnSIE3dZqqog4OnvFPLZ6YTGYshamBwC2t++d02C6dl/Bev4KurAlqJsjWirC7uBUlzbQUHFAE9D8L9TS9jEjWoyJV9Q0P3WyMqfaMl0RK60EYPIfg6ui4SFARqAyrwKbRa0tYVchJOXWzqIp5IYpViIJ8H7h842DOYDiRKAAMoj0h5lFnPLVJc2pQfbEf/8Kt/oOtTzWAAJnynhVab5kC6IPKemIJ2PDrEmKFdZ9IPGilixNqaH94/FdTLBJ8PtWxMnC7MKK85UZSxZfP6hkp7VIZjbHKwUhUhMPoCNrgISSwmVRDw9XIs7bEYTU161sAPreJfwIF1Bep8gH0xwJkszNvH2AR05JCh4uoV6FBOud8Wi/DNEtr3rEIwXtw0J/ShR7K7LEkT2UmleZ97Kfg8B46ckqqNoaXVjFPp1f8x6f5Y2F6hi7W/XjcvBj59Ti26dBpROX9yNiDAH7tRnODJwmQd2FxOMMVhBGFZzpS8di1anFmyPhm7GtgzXL2XMsuniFTSkafe2AQ+7Sem0oCloHIDG5o30ouw/RjY6SPoB0urkxYJPrXDC+9bqeSd/GpiJBptRsHfFtB/VP79GInjsXq9frhcjB2hKbywlBqhXDyeZjDoqAVvMAJv0tS1jydgYc9GWmeUJ/gPtj66E2E9zx2XKksZBoynqLzUEr7bs19L2zQ0Qz2wX+w7c9gQO7tu7IOa4A1XFt87j0Bcj+PUIb6ANDVIbASvlndHvQtBCGNwbA17Drox+tG8y8XAu/y9stAUcm86/tg7pSe8lA+EjdPkCIomGFDsxJbA1ciB4zhMI4ev5iCpOU6jQaHEMjySoruutITIkvwkLmnouajHNV8OeavJRJry3pO4RPi+Sc1MzOLsh5yhike0qUSK6fbFCEp5elqbW3aAw4c1YKPJu6T/p4wdL15LaTaaRo9IpwzygalXk6phbyuSmhgPm2QPeeKdvIqD4tuj+INLgpt6iQffsSehjCwzck5R4MEHYBbJ6T3eCQEds1kjicvlVMYW20bPw8FJdv45jF4pxefuRC6NVtsgNHWBguErBOleLrLg3SlFUohvqBCydzWyfAYpmhUUmai4HQKrNwsSlW+Npbn8rRFSInpGz9ia3Eq5T2e1ZtVqNtBe5MLR/P4RhNTDnBa8mp/idXkVr/QgvqByjJGPgSYSVuhDA4C5UlSWnpM6IjZvgZ+rZZ2tGyxE5gONhvbMlrV2pyuENqF9lZOn7U2v7/GFfOQiSVZRkoc7ACwWgW2qFmgstqPgWUeLz+enj4T13YIMNOVMA0Jhxr3G63rTSPalhXSmjsVm+D+hkrELdmiaPcB/TD9h1/xLTaUxETS/ZnJPPxP6F0Q/yWpDtQoQLmbEyNHdckhyD5RGEH0AHzNP9m/wHwEcfWMSHGLJ9LSOvMbK+V1yEOmBoIuJcm05wtoQUdND7sJj2IE8bRVmth5JOFSMnI6cFiuYhZC3DzWjPpY3Go8Obr4yG95qc54IhAKE7FQfYNvNYv1j2RDmMFrlNQRPAFQFc1EF1HH6QScq1aaMpy+lnCqNREssKbJHu2OnzS8psOl4XPNXHaP9nZ/fcGh931TSJMSvcyufCQ/Jh4skm1NeO9XqXSphQVvWSMs5eZFMmFcFV4neLuAqgayjvCyVY2y+oXfME3DKA/V6lCktov29QrRLy1l8ybosWYqOrpEDsVh8bGaql65kyc0uBN6z1DVz4hF+FCzWr9teUPk1dCbWAOb1rUmB8ZIiSF3UPLSVQ9pa5mygWFnH0FswlUeeV+4xC5Xr20LSvAaCVL8F5cTEgvNqCqbcfD27w1bQsEmFZ4iPx87ajYVwC8zLnoEX24kxDweb5lzONPcP/9Hjsan9M5FmqRLzL7WdL6LnXvP11WoaMUn7Ru3iVTPLaQCaWG0DMeda4C5TeTXInFX4xFeQNuMx71UaHxcL3f+GdLRXsdkUhw6j82rSHUYKsR9F6kq6Y09s1oMDh8pKBJzNiFZu+Y66VoWfZhvl+EjWOn8RJglKC6EFz8xxwWlHOckx4jx+kztOxquOgPh8KqibIoyaKBLqn8vJZKls53TWkSEtaU3OJ/7JYaopPrhkTu80/UmdunMe2cOXObSsvhVckrdQhP5u0yyZhlkLwiw8MJyy2FQIbwlNUIcm0uN+XojoyPDyvjSZ4N2pFea10B3XUhZ9hdTPBPhL55Fn9gILHPXYvQN2dEwyR9pYxHEtjanasoyOoBUePcg66ameaNWUkwZSrc1s0yCRLRF3d4IR+vW//D9oQKWG")).decode("utf-8"))
H = NL.join(a["id"] + " - " + a["название"] for a in D["health_articles"])
L = NL.join(t["id"] + " - " + t["название"] for t in D["law_topics"])
SYS = NL.join([
    "Ты помощник призывника в России. Тебе дают вопрос человека. Определи область:",
    "здоровье - спрашивают, возьмут ли в армию / какая категория годности при болезни, травме, диагнозе (ответ - номер статьи Расписания болезней);",
    "право - призыв, повестки, отсрочки, воинский учёт, документы военкомата, обжалование, медкомиссия как процедура (ответ - id правовой темы);",
    "вне - всё остальное (тогда id пустой).",
    "", "СТАТЬИ РАСПИСАНИЯ БОЛЕЗНЕЙ:", H, "", "ПРАВОВЫЕ ТЕМЫ:", L, "",
    'Ответь ТОЛЬКО одной строкой JSON: {"область": "здоровье|право|вне", "id": "номер статьи или id темы или пусто"}'])
RX = re.compile("[{].*[}]", re.S)

def ask(llm, q):
    t = time.time()
    r = llm.prompt(SYS + NL + NL + "ВОПРОС: " + q)
    dt = time.time() - t
    m = RX.search(str(r))
    try:
        a = json.loads(m.group(0)) if m else {}
    except Exception:
        a = {}
    return a, dt

RES = {}
def score(name, n=50):
    llm = kbench.llms[name]
    ok_area = ok_ans = n_ans = 0
    times, rows = [], []
    for q in D["questions"][:n]:
        try:
            a, dt = ask(llm, q["text"])
        except Exception as e:
            a, dt = {"ошибка": str(e)[:200]}, 0.0
        times.append(dt)
        got = a.get("область", "вне")
        ok_area += (q["area"] == "спорно") or (got == q["area"])
        if q["area"] in ("здоровье", "право") and q["answers"]:
            n_ans += 1
            ok_ans += got == q["area"] and str(a.get("id", "")) in q["answers"]
        rows.append({"i": q["i"], "ответ": a, "сек": round(dt, 2)})
    times.sort()
    RES[name] = {"область": ok_area, "из": n, "ответ": ok_ans, "из_с_ответом": n_ans, "медиана_сек": round(times[len(times) // 2], 2), "строки": rows}
    print(name, "| область", ok_area, "/", n, "| статья/тема", ok_ans, "/", n_ans, "| медиана", RES[name]["медиана_сек"], "с", flush=True)

MODELS = ["google/gemini-3.8-flash", "anthropic/claude-haiku-4-5@20251001", "openai/gpt-6-luna", "anthropic/claude-sonnet-5-5@default"]
for m in MODELS:
    score(m)
json.dump(RES, open("/kaggle/working/nnl_flagships.json", "w"), ensure_ascii=False)
```
